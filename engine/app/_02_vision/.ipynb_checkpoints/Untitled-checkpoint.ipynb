```python
import cv2
import matplotlib.pyplot as plt


img = cv2.imread(image_path)

# Convert to gray
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)


# Blob detector parameters
params = cv2.SimpleBlobDetector_Params()

# Detect dark objects
params.filterByColor = True
params.blobColor = 0

# Area filter
params.filterByArea = True
params.minArea = 20
params.maxArea = 800

# Circularity
params.filterByCircularity = True
params.minCircularity = 0.4

# Convexity
params.filterByConvexity = True
params.minConvexity = 0.5

# Inertia
params.filterByInertia = True
params.minInertiaRatio = 0.3


# Create detector
detector = cv2.SimpleBlobDetector_create(params)


# Detect blobs
keypoints = detector.detect(gray)

print("Detected blobs:", len(keypoints))


# Draw detected blobs
result = cv2.drawKeypoints(
    img,
    keypoints,
    None,
    (0,0,255),
    cv2.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS
)


# Display result
plt.figure(figsize=(7,7))
plt.imshow(cv2.cvtColor(result, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()


# Print coordinates
for i, kp in enumerate(keypoints):
    x, y = kp.pt
    size = kp.size
    print(
        f"{i}: x={x:.1f}, y={y:.1f}, size={size:.1f}"
    )
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt


def minimap_to_world(mx, my, w, h):

    world_x = (15 / w) * mx + (25 / h) * my
    world_y = (15 / w) * mx - (25 / h) * my

    return world_x, world_y
```

```python


img = cv2.imread(image_path)

img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

height, width = img.shape[:2]

w = width / 2
h = height / 2

print("width:", width)
print("height:", height)
print("w:", w)
print("h:", h)
```

```python
points_x = []
points_y = []
colors = []


for y in range(height):

    for x in range(width):

        # 미니맵 중심 기준 좌표
        mx = x - w
        my = y - h


        # 월드 좌표 변환
        wx, wy = minimap_to_world(
            mx,
            my,
            w,
            h
        )

        points_x.append(wx)
        points_y.append(wy)

        colors.append(
            img_rgb[y,x]/255
        )
```

```python
plt.figure(figsize=(8,8))

plt.scatter(
    points_x,
    points_y,
    c=colors,
    s=1
)

plt.xlabel("World X")
plt.ylabel("World Y")

plt.axis("equal")
plt.grid()

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt


def minimap_to_world(mx, my, w, h):

    world_x = (15 / w) * mx + (25 / h) * my
    world_y = (15 / w) * mx - (25 / h) * my

    return world_x, world_y



# 이미지 읽기

img = cv2.imread(image_path)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)


height, width = img.shape[:2]

w = width / 2
h = height / 2



# 변환된 좌표 저장
world_x = []
world_y = []
colors = []


for y in range(height):

    for x in range(width):

        mx = x - w
        my = y - h

        wx, wy = minimap_to_world(
            mx,
            my,
            w,
            h
        )

        world_x.append(wx)
        world_y.append(wy)

        colors.append(img[y,x]/255)



# 월드 좌표계로 표시
plt.figure(figsize=(8,8))


plt.scatter(
    world_x,
    world_y,
    c=colors,
    s=2
)


# 월드 좌표 눈금
plt.xlabel("World X")
plt.ylabel("World Y")

plt.grid(True)

plt.axis("equal")


# 중심 표시
plt.scatter(
    0,
    0,
    c="red",
    s=80
)

plt.text(
    0,
    0,
    " Player (0,0)",
    color="red"
)


plt.show()
```
